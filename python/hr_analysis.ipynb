```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("HR Aanlytics Project Started!")
print("Pandas Version: ", pd.__version__)
```

```python
df = pd.read_csv("../data/HR-Employee-Attrition.csv")
print("Dataset loaded Successfully")
print("Dataset Shape:", df.shape)
display(df.head())
```

```python
df.info()
```

```python
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
display(df.describe())
```

```python
#Checking missing values
print("Total missing values:", df.isnull().sum().sum())

#Checking duplicate rows
print("Duplicate rows:", df.duplicated().sum())

#Checking attrition distribution
print("\nAttrition distribution:")
print(df["Attrition"].value_counts())

#Checking attrition percentages
print("\nAttrition percentage:")
print((df["Attrition"].value_counts(normalize=True) * 100).round(2))

#Checking unique values in every column
print("\nUnique values per column:")
print(df.nunique().sort_values())
```

```python
total_employees = len(df)
employees_left = (df["Attrition"] == "Yes").sum()
employees_stayed = (df["Attrition"] == "No").sum()
attrition_rate = employees_left / total_employees * 100

print("Total Employees:", total_employees)
print("Employees Left:", employees_left)
print("Employees Stayed:", employees_stayed)
print(f"Attrition Rate: {attrition_rate:.2f}%")
```

```python
# attrition by department
department_attrition = (
    df.groupby("Department")
      .agg(
          Total_Employees=("Attrition", "size"),
          Employees_Left=("Attrition",
                          lambda x: (x == "Yes").sum())
      )
)

department_attrition["Attrition_Rate"] = (
    department_attrition["Employees_Left"]
    / department_attrition["Total_Employees"] * 100
).round(2)

display(department_attrition.sort_values(
    "Attrition_Rate", ascending=False
))
```

```python
department_attrition["Attrition_Rate"].sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Employee Attrition Rate by Department")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Department")
plt.tight_layout()
plt.show()
```

```python
overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

display(overtime_attrition.round(2))

overtime_attrition.plot(
    kind="bar",
    figsize=(7, 4)
)

plt.title("Attrition by Overtime Status")
plt.xlabel("Overtime")
plt.ylabel("Percentage of Employees (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Average monthly income by attrition
income_analysis = (
    df.groupby("Attrition")["MonthlyIncome"]
      .agg(["count", "mean", "median"])
      .round(2)
)
display(income_analysis)
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income by Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")
plt.tight_layout()
plt.show()
```

```python
# Employee attrition by age
plt.figure(figsize=(10, 5))
sns.histplot(
    data=df,
    x="Age",
    hue="Attrition",
    multiple="stack",
    bins=15
)

plt.title("Employee Attrition Distribution by Age")
plt.xlabel("Employee Age")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
# age groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"],
    include_lowest=True
)

#attrition by age group
age_attrition = (
    df.groupby("AgeGroup", observed=False)
    .agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
)

age_attrition["Attrition_Rate"] = (
    age_attrition["Employees_Left"]
    / age_attrition["Total_Employees"] * 100
).round(2)

display(age_attrition)
```

```python
# Employee attrition by job role
role_attrition = (
    df.groupby("JobRole")
    .agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
)

role_attrition["Attrition_Rate"] = (
    role_attrition["Employees_Left"]
    / role_attrition["Total_Employees"] * 100
).round(2)

role_attrition = role_attrition.sort_values(
    "Attrition_Rate",
    ascending=False
)

display(role_attrition)

# Visualizing results
plt.figure(figsize=(10, 6))
sns.barplot(
    data=role_attrition.reset_index(),
    x="Attrition_Rate",
    y="JobRole"
)

plt.title("Employee Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")
plt.tight_layout()
plt.show()
```

```python
# Attrition by job satisfaction level
satisfaction_attrition = (
    df.groupby("JobSatisfaction")
    .agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
)

satisfaction_attrition["Attrition_Rate"] = (
    satisfaction_attrition["Employees_Left"]
    / satisfaction_attrition["Total_Employees"] * 100
).round(2)

display(satisfaction_attrition)

# Visualizing attrition rates
plt.figure(figsize=(8, 5))
sns.barplot(
    x=satisfaction_attrition.index,
    y=satisfaction_attrition["Attrition_Rate"]
)
plt.title("Employee Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction Level (1 = Low, 4 = Very High)")
plt.ylabel("Attrition Rate (%)")
plt.tight_layout()
plt.show()
```

```python
# Attrition by work-life balance
worklife_attrition = (
    df.groupby("WorkLifeBalance")
    .agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
)

worklife_attrition["Attrition_Rate"] = (
    worklife_attrition["Employees_Left"]
    / worklife_attrition["Total_Employees"] * 100
).round(2)

display(worklife_attrition)

# visualization
plt.figure(figsize=(8, 5))

sns.barplot(
    x=worklife_attrition.index,
    y=worklife_attrition["Attrition_Rate"]
)

plt.title("Employee Attrition Rate by Work-Life Balance")
plt.xlabel("Work-Life Balance (1 = Poor, 4 = Best)")
plt.ylabel("Attrition Rate (%)")
plt.tight_layout()
plt.show()
```

```python
# Attrition by years at the company
df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 1, 3, 5, 10, 40],
    labels=["0-1 years", "2-3 years", "4-5 years", "6-10 years", "11+ years"
    ]
)

tenure_attrition = (
    df.groupby("TenureGroup", observed=False)
    .agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
)

tenure_attrition["Attrition_Rate"] = (
    tenure_attrition["Employees_Left"]
    / tenure_attrition["Total_Employees"] * 100
).round(2)

display(tenure_attrition)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=tenure_attrition.index,
    y=tenure_attrition["Attrition_Rate"]
)

plt.title("Employee Attrition Rate by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Attrition Rate (%)")
plt.tight_layout()
plt.show()
```

```python
# Employee attrition by percentage salary hike
hike_attrition = (
    df.groupby("PercentSalaryHike")
    .agg(
        Total_Employees=("Attrition", "size"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
)

hike_attrition["Attrition_Rate"] = (
    hike_attrition["Employees_Left"]
    / hike_attrition["Total_Employees"] * 100
).round(2)

display(hike_attrition.sort_values(
    "Attrition_Rate", ascending=False
))

plt.figure(figsize=(10, 5))

sns.lineplot(
    x=hike_attrition.index,
    y=hike_attrition["Attrition_Rate"],
    marker="o"
)

plt.title("Employee Attrition Rate by Percentage Salary Hike")
plt.xlabel("Percentage Salary Hike")
plt.ylabel("Attrition Rate (%)")
plt.xticks(sorted(df["PercentSalaryHike"].unique()))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
